**(a) Explain the problem and why AI is suitable.**

**The Problem: Predicting Food Demand**

CompanyA faces the challenge of accurately predicting the amount of food needed to provide lunchboxes to 40 families for 30 days, especially with an event like Hari Raya approaching. Inaccurate predictions can lead to several issues:

1.  **Food Waste:** Overestimating demand results in surplus food that might spoil, leading to financial loss and ethical concerns about wasted resources.
2.  **Food Shortages:** Underestimating demand means some families might not receive lunchboxes, failing the organization's mission and potentially causing hardship.
3.  **Inefficient Resource Allocation:** Without accurate predictions, CompanyA might misallocate resources such as ingredients, preparation time, volunteer efforts, and transportation, leading to operational inefficiencies and increased costs.
4.  **Impact of Special Events:** Events like Hari Raya can significantly alter demand patterns due to changes in family gatherings, income, or availability of other food sources, making historical averages less reliable.

**Why AI is Suitable for This Problem**

Artificial Intelligence is an excellent fit for solving this food demand prediction problem due to its capabilities:

1.  **Handling Complex Data:** Food demand is influenced by numerous factors, including historical consumption patterns, seasonal changes, holidays (like Hari Raya), local events, economic conditions, and even weather. AI models can process and identify intricate relationships within this diverse and complex dataset much more effectively than traditional statistical methods.
2.  **Improved Prediction Accuracy:** AI algorithms, particularly machine learning techniques, can learn from past data to build predictive models that offer higher accuracy. This minimizes both overstocking (reducing waste) and understocking (ensuring all families receive food).
3.  **Adaptability and Learning:** AI models can be continuously updated and retrained with new data. This means they can adapt to evolving demand patterns, incorporate the impact of new holidays or economic shifts, and improve their predictions over time as more data becomes available.
4.  **Pattern Recognition:** AI can detect subtle patterns and trends that might be invisible to human analysts or simpler models. For instance, it can discern how Hari Raya historically affects demand, allowing for better planning.
5.  **Optimization:** By providing accurate forecasts, AI enables CompanyA to optimize its entire supply chain—from procurement of ingredients to the scheduling of preparation and delivery—leading to greater efficiency, cost savings, and ultimately, better service to the community.

```python
import numpy as np
import pandas as pd

# -----------------------------
# Updated assumptions for 40 families with variable appetite
# -----------------------------
num_days = 30
num_families = 40
mean_price = 10.0           # Average Malaysia lunchbox price (RM)
price_std = 1.0             # Price fluctuation

np.random.seed(42)

# 1. Simulate fixed family sizes (base capacity)
family_sizes = np.random.randint(2, 6, size=num_families)
base_daily_total = family_sizes.sum()

# 2. Simulate daily appetite fluctuation (e.g., +/- 10% of base)
# This accounts for the fact that demand isn't identical every day
appetite_fluctuation = np.random.randint(-0, 16, size=num_days)
daily_lunchboxes = base_daily_total + appetite_fluctuation

# 3. Simulate daily price fluctuation
daily_price = np.random.normal(mean_price, price_std, num_days)
daily_price = np.clip(daily_price, 8, None)

# 4. Calculate daily and total cost
daily_cost = daily_lunchboxes * daily_price
total_project_cost = daily_cost.sum()

# Create results table
prediction = pd.DataFrame({
    "Day": range(1, num_days + 1),
    "Base Capacity": base_daily_total,
    "Actual Lunchboxes (Appetite Adjusted)": daily_lunchboxes,
    "Price per Lunchbox (RM)": np.round(daily_price, 2),
    "Total Daily Cost (RM)": np.round(daily_cost, 2)
})

display(prediction)
print(f"\n--- Final Summary ---")
print(f"Base daily requirement (40 families): {base_daily_total}")
print(f"Total Lunchboxes predicted over 30 days: {daily_lunchboxes.sum()}")
print(f"Estimated Final Project Cost: RM {total_project_cost:,.2f}")
```
